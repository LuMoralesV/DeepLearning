```python
import heapq
from itertools import count

GOAL = (1, 2, 3, 4, 5, 6, 7, 8, 0)

def neighbors(state):
    i = state.index(0)
    r, c = divmod(i, 3)
    moves = []
    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            j = nr * 3 + nc
            s = list(state)
            s[i], s[j] = s[j], s[i]
            moves.append(tuple(s))
    return moves

def manhattan(state):
    dist = 0
    for idx, val in enumerate(state):
        if val == 0: continue
        gr, gc = divmod(val - 1, 3)
        r, c = divmod(idx, 3)
        dist += abs(gr - r) + abs(gc - c)
    return dist

def reconstruct(parent, state):
    path = [state]
    while parent[path[-1]] is not None:
        path.append(parent[path[-1]])
    return list(reversed(path))

# --- BFS ---
def bfs(start):
    frontier, visited = [start], {start}
    parent = {start: None}
    expanded = 0
    while frontier:
        state = frontier.pop(0)
        expanded += 1
        if state == GOAL:
            return reconstruct(parent, state), expanded
        for nxt in neighbors(state):
            if nxt not in visited:
                visited.add(nxt)
                parent[nxt] = state
                frontier.append(nxt)
    return None, expanded

# --- DFS ---
def dfs(start):
    frontier, visited = [start], {start}
    parent = {start: None}
    expanded = 0
    while frontier:
        state = frontier.pop()
        expanded += 1
        if state == GOAL:
            return reconstruct(parent, state), expanded
        for nxt in neighbors(state):
            if nxt not in visited:
                visited.add(nxt)
                parent[nxt] = state
                frontier.append(nxt)
    return None, expanded

# --- A* ---
def astar(start):
    counter = count()
    frontier = [(manhattan(start), next(counter), start, 0)]
    parent, g_score = {start: None}, {start: 0}
    expanded = 0
    while frontier:
        _, _, state, g = heapq.heappop(frontier)
        expanded += 1
        if state == GOAL:
            return reconstruct(parent, state), expanded
        for nxt in neighbors(state):
            new_g = g + 1
            if nxt not in g_score or new_g < g_score[nxt]:
                g_score[nxt] = new_g
                parent[nxt] = state
                heapq.heappush(frontier, (new_g + manhattan(nxt), next(counter), nxt, new_g))
    return None, expanded

# --- Casos de Prueba del Laboratorio ---
test_cases = [
    (1, 7, 6, 2, 0, 8, 4, 5, 3),
    (1, 6, 4, 8, 5, 7, 2, 0, 3),
    (5, 3, 1, 4, 0, 8, 2, 6, 7),
    (2, 1, 4, 3, 5, 6, 7, 0, 8),
    (0, 6, 2, 5, 3, 8, 7, 4, 1)
]

# --- Ejecución e Impresión ---
for i, test in enumerate(test_cases, 1):
    print("=" * 70)
    print(f"EVALUANDO CASO DE PRUEBA {i}: {test}")
    print("=" * 70)
    print("\n--- RESUMEN DEL CASO ---")

    path_bfs, exp_bfs = bfs(test)
    len_bfs = len(path_bfs) - 1 if path_bfs else "N/A"
    print(f"BFS    -> Longitud de ruta: {len_bfs}, Nodos expandidos: {exp_bfs}")

    path_dfs, exp_dfs = dfs(test)
    len_dfs = len(path_dfs) - 1 if path_dfs else "N/A"
    print(f"DFS    -> Longitud de ruta: {len_dfs}, Nodos expandidos: {exp_dfs}")

    path_astar, exp_astar = astar(test)
    len_astar = len(path_astar) - 1 if path_astar else "N/A"
    print(f"A*     -> Longitud de ruta: {len_astar}, Nodos expandidos: {exp_astar}\n")
```
